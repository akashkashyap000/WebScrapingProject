```python
def check_price():
    import requests
    from bs4 import BeautifulSoup
    import datetime
    import csv
    import os

    URL = "https://www.amazon.in/Lux-Cozi-T-Shirt-Pocket_COZI_2122_EMRAD_L_1PC-Emerald/dp/B0CVXRSVSV/ref=sr_1_8?crid=30SB874UI5LC7&dib=eyJ2IjoiMSJ9.WJr6i8-Mn68APCJNUsmFXX_Ms-toE8vkUHuof1rbJU6SFHkBdpoazqD_nrvxKWv02gdSfLD7UjPweUkcIbznHM7XSTEYCCd9gnSbLitPnL2oTLpi5o4Pp4zo8kla_6QqkyUZKSF6jxOUL0A1choGBYV-QLd0QqyUIbflL988r4QbRQppA35x-_MG5sq1yk46wGaag10FXldk2cwS2CGXbTcW87xahPPpT_rjl5K8b92I4whfFeU5iFkabUsHDMxvT5R1YNtRWdeChQc-JmMfzVfl3DAbWGSaAHeZP70TOmw.k76Ope7dwffwt_-N4bRWDxl-bmAew7oRZT2VRl3JLz0&dib_tag=se&keywords=t%2Bshirt&qid=1775457381&sprefix=tshi%2Caps%2C1358&sr=8-8&th=1&psc=1"

    headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/78.0.3904.108 Safari/537.36", "Accept-Encoding":"gzip, deflate", "Accept":"text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8", "DNT":"1","Connection":"close", "Upgrade-Insecure-Requests":"1"}

    page = requests.get(URL, headers=headers)
    soup = BeautifulSoup(page.content, "html.parser")

    
    title_tag = soup.find(id='productTitle')
    if title_tag:
        title = title_tag.get_text().strip()
    else:
        title = "Not Found"

    # Price (updated selector)
    price_tag = soup.find('span', class_='a-price-whole')
    if price_tag:
        price = price_tag.get_text().strip()
    else:
        price = "Not Found"

    # Date
    today = datetime.date.today()

    # File handling
    file_name = 'AmazonWebScraperDataset.csv'
    header = ['Title', 'Price', 'Date']
    data = [title, price, today]

    # Write header only if file doesn't exist
    file_exists = os.path.isfile(file_name)

    with open(file_name, 'a', newline='', encoding='UTF8') as f:
        writer = csv.writer(f)
        if not file_exists:
            writer.writerow(header)
        writer.writerow(data)

    print("Data Saved ")
    print(title)
    print(price)
```

```python
# Runs check_price after a set time and inputs data into your CSV

while(True):
    check_price()
    time.sleep(86400)
```
